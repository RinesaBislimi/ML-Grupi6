```python
numeric_columns = dataset.select_dtypes(include=['int64', 'float64']).columns

# Initialize a dictionary to store the count of outliers per column
outliers = {}

# Iterate over each numeric column to calculate the IQR and detect outliers
for col in numeric_columns:
    Q1 = dataset[col].quantile(0.25)
    Q3 = dataset[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Get the count of outliers in the column
    outliers_count = dataset[(dataset[col] < lower_bound) | (dataset[col] > upper_bound)].shape[0]
    
    # Store the count of outliers in the dictionary
    outliers[col] = outliers_count

# Display the results
outliers_df = pd.DataFrame(list(outliers.items()), columns=['Column', 'Outliers Count'])
outliers_df
```

```python
def calculate_heart_attack_risk(row):
    risk_score = 0
    
    # LDL Level Risk
    if row['LDL_Level'] >= 160:
        risk_score += 2
    elif row['LDL_Level'] >= 130:
        risk_score += 1
    
    # Triglyceride Level Risk
    if row['Triglyceride_Level'] >= 200:
        risk_score += 1

    # Systolic BP Risk
    if row['Systolic_BP'] >= 140:
        risk_score += 2
    elif row['Systolic_BP'] >= 130:
        risk_score += 1
    
    # Diabetes Risk
    if row['Diabetes'] == 1:
        risk_score += 2
    
    # Smoking Risk
    if row['Smoking'] == 1:
        risk_score += 1
    
    # Physical Activity Risk
    if row['Physical_Activity'] <= 4:
        risk_score += 1

    # Family History Risk
    if row['Family_History'] == 1:
        risk_score += 1
    
    # Age Risk
    if row['Age'] >= 60:
        risk_score += 2
    elif row['Age'] >= 45:
        risk_score += 1

    # Stress Level Risk (new addition)
    if row['Stress_Level'] >= 7:
        risk_score += 2
    elif row['Stress_Level'] >= 4:
        risk_score += 1
    
    return risk_score

# Apply the updated risk calculation function to the dataset
dataset['Heart_Attack_Risk_Score'] = dataset.apply(calculate_heart_attack_risk, axis=1)

# Function to categorize risk based on the score
def risk_category(score):
    if score >= 6:
        return "High Risk"
    elif score >= 4:
        return "Moderate Risk"
    else:
        return "Low Risk"

# Assign heart attack risk category based on the score
dataset['Heart_Attack_Risk_Category'] = dataset['Heart_Attack_Risk_Score'].apply(risk_category)

# Display the selected columns
dataset[['Patient_ID', 'Heart_Attack_Risk_Score', 'Heart_Attack_Risk_Category']].head()

# Group by Heart Attack Risk Category and calculate mean values for other factors
risk_summary = dataset.groupby('Heart_Attack_Risk_Category')[['Age', 'HDL_Level', 'LDL_Level', 'Diastolic_BP', 'Systolic_BP', 'Stress_Level']].mean()

# Print the risk summary
print(risk_summary)

```
